```python
# EX No: 1A
# F1 RACE ANALYTICS – JUPYTER NOTEBOOK

# Program 1 – Simple Python Program
print("Welcome to F1 Race Analytics Lab")

# Program 2 – Import Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully")

# Program 3 – Load F1 Results Dataset
df = pd.read_csv("results.csv")

print("\nF1 Race Results Dataset:")
print(df.head())

# Program 4 – Display Dataset Information
print("\nDataset Information:")
df.info()

print("\nDataset Shape:")
print(df.shape)

# Program 5 – Display Column Names
print("\nColumn Names:")
print(df.columns.tolist())

# Program 6 – Basic Statistical Analysis
print("\nStatistical Summary:")
print(df.describe())

# Program 7 – Total Points by Driver
driver_points = df.groupby("driverId")["points"].sum()

top_10_drivers = driver_points.sort_values(
    ascending=False
).head(10)

print("\nTop 10 Drivers by Total Points:")
print(top_10_drivers)

# Program 8 – Grid Position vs Finishing Position
performance = df[["driverId", "grid", "position"]].copy()

performance["position"] = pd.to_numeric(
    performance["position"],
    errors="coerce"
)

performance = performance.dropna()

performance["position_gain"] = (
    performance["grid"] - performance["position"]
)

print("\nGrid Position vs Finishing Position:")
print(performance.head())

# Program 9 – Data Visualization
top_10_drivers.sort_values().plot(
    kind="barh",
    figsize=(8, 5)
)

plt.title("Top 10 F1 Drivers by Total Points")
plt.xlabel("Total Points")
plt.ylabel("Driver ID")
plt.show()
```
