```python
# Import libraries

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

# Load F1 dataset
df = pd.read_csv("results.csv")

# Convert important columns to numeric
numeric_columns = [
    "grid",
    "position",
    "points",
    "laps",
    "fastestLap",
    "rank",
    "fastestLapSpeed"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Dataset Preview
print("Dataset Preview:")
print(df.head())

# Dataset Shape
print("\nDataset Shape:", df.shape)

# Statistical Summary
print("\nStatistical Summary:")
print(df.describe())

# Filter F1 race records
filtered = df[df["points"] > df["points"].mean()]

print("\nRace Records with Points Above Average:")
print(filtered.head())

# Statistical Analysis
points = df["points"]

print("\nMean:", points.mean())
print("Median:", points.median())
print("Minimum:", points.min())
print("Maximum:", points.max())
print("Standard Deviation:", points.std())

# Identify numerical columns
num = df.select_dtypes(include=np.number)

print("\nNumerical Columns:")
print(num.columns.tolist())

# Correlation Matrix
print("\nCorrelation Matrix:")
print(num.corr())

# Correlation Heatmap
plt.figure(figsize=(10, 7))

sns.heatmap(
    num.corr(),
    annot=True,
    cmap="Blues",
    fmt=".2f"
)

plt.title("Correlation Analysis of F1 Race Features")
plt.show()
```
