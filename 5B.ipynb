```python
# EXPT NO: 5B
# DATA VISUALIZATION – F1 RACE ANALYTICS

import pandas as pd
import matplotlib.pyplot as plt

# Load F1 Race Results Dataset
df = pd.read_csv("results.csv")

# Convert required columns to numeric
df["grid"] = pd.to_numeric(df["grid"], errors="coerce")
df["position"] = pd.to_numeric(df["position"], errors="coerce")
df["points"] = pd.to_numeric(df["points"], errors="coerce")

# Display dataset
print("F1 Race Results Dataset:")
print(df.head())

# 1. BAR CHART – Driver-wise Total Points
driver_points = (
    df.groupby("driverId")["points"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(8, 5))
plt.bar(
    driver_points.index.astype(str),
    driver_points.values
)
plt.title("Top 10 F1 Drivers by Total Points")
plt.xlabel("Driver ID")
plt.ylabel("Total Points")
plt.show()

# 2. LINE CHART – Race Points Trend
plt.figure(figsize=(8, 5))
plt.plot(
    df.head(30)["resultId"],
    df.head(30)["points"],
    marker="o"
)
plt.title("F1 Race Points Trend")
plt.xlabel("Result ID")
plt.ylabel("Points")
plt.show()

# 3. PIE CHART – Top 5 Drivers
top_5 = (
    df.groupby("driverId")["points"]
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

plt.figure(figsize=(7, 7))
plt.pie(
    top_5.values,
    labels=top_5.index.astype(str),
    autopct="%1.1f%%"
)
plt.title("Top 5 Drivers – Points Distribution")
plt.show()

# 4. HISTOGRAM – Points Distribution
plt.figure(figsize=(8, 5))
plt.hist(
    df["points"].dropna(),
    bins=10
)
plt.title("Distribution of F1 Race Points")
plt.xlabel("Points")
plt.ylabel("Frequency")
plt.show()

# 5. SCATTER PLOT – Grid Position vs Finishing Position
plt.figure(figsize=(8, 5))
plt.scatter(
    df["grid"],
    df["position"]
)
plt.title("Grid Position vs Finishing Position")
plt.xlabel("Grid Position")
plt.ylabel("Finishing Position")
plt.show()

# 6. BOX PLOT – Points Variation
plt.figure(figsize=(6, 5))
plt.boxplot(
    df["points"].dropna()
)
plt.title("F1 Race Points Box Plot")
plt.ylabel("Points")
plt.show()

print("\nAll F1 Data Visualizations Completed Successfully")
```
