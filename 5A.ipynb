```python
# EXPT NO: 5A
# DATA VISUALIZATION USING MATPLOTLIB – F1 RACE DATA

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load F1 Race Results Dataset
df = pd.read_csv("results.csv")

# Convert required columns to numeric
df["points"] = pd.to_numeric(df["points"], errors="coerce")
df["grid"] = pd.to_numeric(df["grid"], errors="coerce")
df["position"] = pd.to_numeric(df["position"], errors="coerce")
df["fastestLapSpeed"] = pd.to_numeric(
    df["fastestLapSpeed"], errors="coerce"
)

print("F1 Race Results Dataset:")
print(df.head())

# --------------------------------------------------
# PROGRAM 1: LINE CHART
# --------------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    df.head(50)["resultId"],
    df.head(50)["points"],
    marker="o"
)

plt.title("F1 Race Points Trend")
plt.xlabel("Result ID")
plt.ylabel("Points")
plt.show()


# --------------------------------------------------
# PROGRAM 2: BAR CHART
# --------------------------------------------------

driver_points = (
    df.groupby("driverId")["points"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(8, 5))

plt.bar(
    driver_points.index.astype(str),
    driver_points.values
)

plt.title("Top 10 F1 Drivers by Total Points")
plt.xlabel("Driver ID")
plt.ylabel("Total Points")
plt.show()


# --------------------------------------------------
# PROGRAM 3: PIE CHART
# --------------------------------------------------

top_5 = (
    df.groupby("driverId")["points"]
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

plt.figure(figsize=(7, 7))

plt.pie(
    top_5.values,
    labels=top_5.index.astype(str),
    autopct="%1.1f%%"
)

plt.title("Top 5 Drivers – Points Distribution")
plt.show()


# --------------------------------------------------
# PROGRAM 4: HISTOGRAM
# --------------------------------------------------

plt.figure(figsize=(8, 5))

plt.hist(
    df["points"].dropna(),
    bins=10
)

plt.title("F1 Race Points Distribution")
plt.xlabel("Points")
plt.ylabel("Frequency")
plt.show()


# --------------------------------------------------
# PROGRAM 5: SCATTER PLOT
# --------------------------------------------------

scatter_data = df[["grid", "position"]].dropna()

plt.figure(figsize=(8, 5))

plt.scatter(
    scatter_data["grid"],
    scatter_data["position"]
)

plt.title("Grid Position vs Finishing Position")
plt.xlabel("Grid Position")
plt.ylabel("Finishing Position")
plt.show()


# --------------------------------------------------
# PROGRAM 6: BOX PLOT
# --------------------------------------------------

plt.figure(figsize=(6, 5))

plt.boxplot(
    df["points"].dropna()
)

plt.title("F1 Race Points Box Plot")
plt.ylabel("Points")
plt.show()


# --------------------------------------------------
# PROGRAM 7: MULTIPLE LINE CHART
# --------------------------------------------------

line_data = df[
    ["resultId", "points", "fastestLapSpeed"]
].dropna().head(50)

plt.figure(figsize=(9, 5))

plt.plot(
    line_data["resultId"],
    line_data["points"],
    label="Points"
)

plt.plot(
    line_data["resultId"],
    line_data["fastestLapSpeed"],
    label="Fastest Lap Speed"
)

plt.title("F1 Points and Fastest Lap Speed")
plt.xlabel("Result ID")
plt.ylabel("Value")
plt.legend()

plt.show()

print("\nAll F1 Data Visualizations Completed Successfully")
```
